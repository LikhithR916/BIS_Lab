```python
import numpy as np

# Warehouse locations
# 0 = Packing station
locations = np.array([
    [0, 0],
    [2, 3],
    [5, 4],
    [6, 1],
    [3, 7],
    [8, 5]
])

N = len(locations)

# ACO parameters
ANTS = 10
ITERATIONS = 100
ALPHA = 1
BETA = 2
EVAPORATION = 0.5
Q = 100

# Calculate distance matrix
distance = np.zeros((N, N))

for i in range(N):
    for j in range(N):
        distance[i][j] = np.linalg.norm(
            locations[i] - locations[j]
        )

# Initialize pheromone
pheromone = np.ones((N, N))

best_route = None
best_distance = float('inf')

# Print input
print("Warehouse Locations:")
for i, loc in enumerate(locations):
    if i == 0:
        print("Packing Station:", loc)
    else:
        print("Shelf", i, ":", loc)

print("\nDistance Matrix:")
print(np.round(distance, 2))

print("\nACO Parameters:")
print("Number of Ants:", ANTS)
print("Iterations:", ITERATIONS)
print("Alpha:", ALPHA)
print("Beta:", BETA)
print("Evaporation Rate:", EVAPORATION)

# ACO algorithm
for iteration in range(ITERATIONS):

    routes = []
    route_distances = []

    for ant in range(ANTS):

        route = [0]
        unvisited = list(range(1, N))

        while unvisited:

            current = route[-1]
            probabilities = []

            for node in unvisited:
                value = (
                    pheromone[current][node] ** ALPHA
                    * (1 / distance[current][node]) ** BETA
                )
                probabilities.append(value)

            probabilities = np.array(probabilities)
            probabilities /= probabilities.sum()

            next_node = np.random.choice(
                unvisited, p=probabilities
            )

            route.append(next_node)
            unvisited.remove(next_node)

        route.append(0)

        total_distance = sum(
            distance[route[i]][route[i + 1]]
            for i in range(len(route) - 1)
        )

        routes.append(route)
        route_distances.append(total_distance)

        if total_distance < best_distance:
            best_distance = total_distance
            best_route = route.copy()

    # Evaporate pheromones
    pheromone *= (1 - EVAPORATION)

    # Deposit pheromones
    for route, dist in zip(routes, route_distances):

        deposit = Q / dist

        for i in range(len(route) - 1):
            a = route[i]
            b = route[i + 1]

            pheromone[a][b] += deposit
            pheromone[b][a] += deposit

# Print final results
print("\n--- ACO Optimization Results ---")

print("Best Picking Route:")
print(" -> ".join(map(str, best_route)))

print("Minimum Picking Distance:",
      round(best_distance, 2))

```
